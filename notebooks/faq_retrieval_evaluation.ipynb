# Multilingual FAQ retrieval evaluation

This notebook evaluates the deterministic retriever against the transparent test set in `examples/evaluation_queries.json`. It is a development diagnostic, not a production-accuracy claim.

```python
import json
from pathlib import Path
from taja_bot.models import LanguageCode
from taja_bot.services.faq import FaqService

root = Path('..').resolve()
service = FaqService(root / 'src/taja_bot/data/faqs.json')
queries = json.loads((root / 'examples/evaluation_queries.json').read_text(encoding='utf-8'))
```

```python
results = []
for item in queries:
    match = service.match(item['query'], LanguageCode(item['language']))
    results.append({**item, 'predicted': match.faq['id'], 'score': match.score})

accuracy = sum(row['predicted'] == row['expected_faq_id'] for row in results) / len(results)
accuracy
```
